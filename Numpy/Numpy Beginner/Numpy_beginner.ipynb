```python
import numpy as np
print(f"Version: {np.__version__}")
```

```python
list1 = [1,2,3]
array1 = np.array(list1)

print(f'Type: {type(array1)}')
```

```python
matris = np.array([[1,2,3],[4,5,6]])
print(matris)
print(f"Dimension of matris: {matris.ndim}.")
```

```python
print("shape :", matris.shape)
print("ndim  :", matris.ndim)
print("size  :", matris.size)
print("dtype :", matris.dtype)
print("itemsize:", matris.itemsize, "byte")
```

```python
zeros = np.zeros((2,3))
print(zeros , '\n')

ones = np.ones((2,3))
print(ones , '\n')

full = np.full((2,3), 5)
print(full, '\n')

eye = np.eye(3) #3x3
print(eye, '\n')
```

```python
condition = array1 > 1
print(condition)

print(array1[condition]) #only true

print(array1[array1 >= 1])
```

```python
array2 = np.arange(12)
print(array2, '\n\n')

array3 = array2.reshape(3,4)
print(array3, '\n\n')

print(array2.reshape(2,3,2), '\n\n')

print(array3.T, '\n\n')

print(array3.flatten())
```

```python
arrayA = np.array([[1,2,3],[4,5,6]])
arrayB = np.array([[7,8,9],[10,11,12]])

print("Toplama   :", arrayA + arrayB)
print("Çıkarma   :", arrayA - arrayB)
print("Çarpma    :", arrayA * arrayB)
print("Bölme     :", arrayA / arrayB)
print("Üs alma   :", arrayA ** arrayB)
```

```python
matrisA = np.array([[1,2,3],[4,5,6]])
arrayC = np.array([100,200,300])

print(matrisA + arrayC, '\n\n')

print(arrayC * 5, '\n\n')

print(matrisA + 5, '\n\n')
```

```python
print(arrayC.sum()) #toplam
print(arrayC.mean()) #ortalama
print(arrayC.max()) #maximum değer
print(arrayC.min()) #minimum değer
print(arrayC.std()) #standart sapma
print(arrayC.var()) #varyans

print(matrisA.sum(axis=0)) #sütun toplamları
print(matrisA.sum(axis=1)) #satır toplamları
```

```python
np.random.seed(42) #tekrar çalıştırdığımızda sayılar değşmez

rastgele_array = np.random.rand(3, 3)      # 0 ile 1 arasında, 3x3 boyutunda rastgele ondalıklı sayılar üretir
print("rand:\n", rastgele_array)


tam_sayilar = np.random.randint(1, 100, size=5)  # 1 (dahil) ile 100 (hariç) arasında 5 rastgele tam sayı üretir
print("randint:", tam_sayilar)

normal_dagilim = np.random.randn(4)   # standart normal dağılımdan (ortalama 0, std 1) 4 sayı üretir
print("randn:", normal_dagilim)
```

```python
firstMatris = np.array([[1,2],[4,5]])
secondMatris = np.array([[7,8],[10,11]])

matrisMult = firstMatris @ secondMatris #Matris çarpımı
print(matrisMult, '\n\n')


determinant = np.linalg.det(firstMatris) #Determinant hesaplar
print(determinant, '\n\n')

inverse = np.linalg.inv(firstMatris) #Tersini alır
print(inverse, '\n\n')

print(firstMatris.T) #Transpose
```
