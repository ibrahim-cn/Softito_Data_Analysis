```python
df_first = df[df["Calories"] == 300.0]
df_second = df[df["Duration"] == 45]
df_united = pd.concat([df_first, df_second]) #list olarak aldıklarını birleştirir.
df_united
```

```python
durations = pd.DataFrame({
    "Duration": df["Duration"].dropna().unique()
})

durations['duration'] = durations['Duration'].apply(lambda x: x/100)

df_area = pd. merge(df,durations,on="Duration") #verilen kolonda tabloları birleştirir
df_area
```

```python
calorie_summary_to_Date = df.groupby("Date").agg(
    total_calories = ("Calories", "sum"),
    average_calories = ("Calories", "mean"),
    max_calories = ("Calories", "max"),
    min_calories = ("Calories", "min"),
    count_calories = ("Calories", "count")
)
print(calorie_summary_to_Date)
print("ortalama kalorisi en fazla olan gün: ", calorie_summary_to_Date["average_calories"].idxmax()) #en fazla olanı bulur
print("ortalama kalorisi en düşük olan gün: ", calorie_summary_to_Date["average_calories"].idxmin()) #en az olanı bulur
```
