```python
def clean_data(x):
  df = x.copy()
  df["BusinessEntityID"] = df["BusinessEntityID"].fillna("unknown") #eksikleri doldur
  df["ModifiedDate"] = pd.to_datetime(df["ModifiedDate"], errors="coerce") #tarih dönüşümü yap.
  # Mantık hatasını düzelt: Yılı 2027'den büyük olan satırlardaki 'Modified Year' değerini NaN yap.
  df.loc[df["Modified Year"] > 2027, "Modified Year"] = np.nan
  #tekrar eden kayıtlar veriyi bozuyorsa düzenle.
  df = df.reset_index(drop=True) #indexi sıfırla.
  return df

df_result = clean_data(df_first)
df_result.head()
```
